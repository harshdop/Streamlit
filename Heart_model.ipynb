```python
import pandas as pd
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report, precision_score, recall_score, f1_score
```

```python
df = pd.read_csv('/home/harsh/Documents/jNotebook/Machine Learning/Logistic Regression Classification/heart.csv')
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
```

```python
df['oldpeak'] = df['oldpeak'] * 10
```

```python
x= df.drop(columns = ['target'])
```

```python
y= df['target']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size= 0.2, random_state= 4)
```

```python
scaler = StandardScaler()
```

```python
x_train = scaler.fit_transform(x_train)
```

```python
x_test = scaler.transform(x_test)
```

```python
model = LogisticRegression()
```

```python
model.fit(x_train,y_train)
```

```python
y_train_pred = model.predict(x_train)
y_test_pred = model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
print(f'Training data accuracy = {accuracy_score(y_train, y_train_pred)}')
```

```python
print(f'Training data accuracy = {accuracy_score(y_test, y_test_pred)}')
```

```python
confusion_matrix(y_test, y_test_pred)
```

```python
#recall score - out of all actual YES cases, how many were predicted correctly
recall_score(y_test, y_test_pred)
```

```python
accuracy_score(y_test, y_test_pred)
```

```python
precision_score(y_test, y_test_pred) #out of all the predicted yes, how many yes we correct
```
